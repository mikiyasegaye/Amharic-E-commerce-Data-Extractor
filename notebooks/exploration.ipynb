```python
import pandas as pd
import sys
from pathlib import Path

# Add the root directory to the system path
sys.path.append(str(Path().resolve().parent))

from preprocessing.amharic_text_cleaner import clean_amharic_text
```

```python
input_path = "../data/raw/telegram_data.xlsx"

df = pd.read_excel(input_path)
df = df[['Channel Title', 'Message']].dropna().copy()
df.head()
```

```python
df['Cleaned_Message'] = df['Message'].astype(str).apply(clean_amharic_text)
df[['Message', 'Cleaned_Message']].head(10)
```

```python
output_path = '../data/processed/cleaned_telegram_data.csv'

df.to_csv(output_path, index=False, encoding='utf-8')
print(f"[✓] Cleaned data saved to {output_path}")
```
