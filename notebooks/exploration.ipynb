```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: load cleaned data
df = pd.read_csv("../data/processed/cleaned_telegram_data.csv")
df = df[['Channel Title', 'Cleaned_Message']].dropna().copy()
df.head()
```

```python
# Compute message length
df['message_length'] = df['Cleaned_Message'].apply(lambda x: len(str(x).strip()))

# Filter out empty or invalid messages
filtered = df[df['message_length'] > 0]

# Plot safe histogram
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 4))
sns.histplot(filtered['message_length'], bins=40, kde=False)
plt.title("Distribution of Message Lengths")
plt.xlabel("Number of Characters")
plt.ylabel("Frequency")
plt.show()
```

```python
df['token_count'] = df['Cleaned_Message'].str.split().apply(len)

plt.figure(figsize=(10, 4))
sns.boxplot(x=df['token_count'])
plt.title("Token Count per Message (Box Plot)")
plt.xlabel("Number of Tokens")
plt.show()
```

```python
df['token_count'] = df['Cleaned_Message'].str.split().apply(len)

plt.figure(figsize=(10, 4))
sns.boxplot(x=df['token_count'])
plt.title("Token Count per Message (Box Plot)")
plt.xlabel("Number of Tokens")
plt.show()
```

```python
top_channels = df['Channel Title'].value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_channels.values, y=top_channels.index)
plt.title("Top Channels by Message Count")
plt.xlabel("Message Count")
plt.ylabel("Channel")
plt.show()
```
